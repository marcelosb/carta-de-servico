```python
# ============================================================
# OMNIVOICE - INICIALIZAÇÃO COMPLETA
# Google Colab + GPU T4 + Voice Clone + Gradio
# EXECUTE APENAS ESTA CÉLULA
# ============================================================

import os
import subprocess
import sys

print("====================================================")
print("🚀 INICIANDO OMNIVOICE")
print("====================================================")

# ------------------------------------------------------------
# 1. INSTALAR / CORRIGIR DEPENDÊNCIAS
# ------------------------------------------------------------

print("\n📦 Verificando dependências...")

subprocess.run(
    [
        sys.executable,
        "-m",
        "pip",
        "install",
        "-q",
        "setuptools<82",
        "jedi>=0.16",
        "omnivoice==0.2.1",
    ],
    check=True,
)

print("✅ Dependências prontas.")

# ------------------------------------------------------------
# 2. VERIFICAR GPU
# ------------------------------------------------------------

import torch

print("\n🖥️ Verificando GPU...")

if not torch.cuda.is_available():
    raise RuntimeError(
        "\n❌ GPU não encontrada!\n\n"
        "No Google Colab vá em:\n"
        "Ambiente de execução → Alterar tipo de ambiente de execução → T4 GPU\n"
        "Depois execute esta célula novamente."
    )

print("✅ CUDA disponível")
print("GPU:", torch.cuda.get_device_name(0))
print("PyTorch:", torch.__version__)

# ------------------------------------------------------------
# 3. CARREGAR OMNIVOICE + ASR
# ------------------------------------------------------------

print("\n====================================================")
print("🧠 CARREGANDO OMNIVOICE")
print("====================================================")
print()
print("Na primeira execução da nova máquina do Colab")
print("serão baixados os modelos automaticamente.")
print()

from omnivoice import OmniVoice

model = OmniVoice.from_pretrained(
    "k2-fsa/OmniVoice",
    device_map="cuda:0",
    dtype=torch.float16,

    # IMPORTANTE:
    # necessário para usar Voice Clone com
    # transcrição automática da voz de referência
    load_asr=True,
)

print("\n✅ OmniVoice carregado!")
print("Sample rate:", model.sampling_rate)

gpu_memory = torch.cuda.memory_allocated() / (1024 ** 3)

print(
    "Memória GPU utilizada:",
    round(gpu_memory, 2),
    "GB"
)

# ------------------------------------------------------------
# 4. ABRIR INTERFACE OFICIAL GRADIO
# ------------------------------------------------------------

print("\n====================================================")
print("🌐 INICIANDO INTERFACE GRADIO")
print("====================================================")

from omnivoice.cli.demo import build_demo

# Fecha uma interface antiga caso a célula
# tenha sido executada novamente na mesma sessão
try:
    demo.close()
except:
    pass

demo = build_demo(
    model,
    "k2-fsa/OmniVoice"
)

print()
print("✅ Interface preparada.")
print()
print("Aguarde aparecer:")
print()
print("Running on public URL: https://xxxxx.gradio.live")
print()
print("👉 Clique no endereço gradio.live")
print()

demo.queue().launch(
    server_name="0.0.0.0",
    server_port=7860,
    share=True,
    show_error=True,
)
```
